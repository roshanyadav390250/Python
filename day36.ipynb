```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
print(tips.head())
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.show()
```

```python
sns.barplot(data=tips,x='sex',y='tip')
```

```python
np.random.seed(42)
n = 100
cust_data = pd.DataFrame({
    'name': [f'Cust_{i}'for i in range(n)],
    'department': np.random.choice(['marketing','IT','sales','finance','HR','accountant'], n),
    'gender': np.random.choice(['Male','Female'],n),
    'age': np.random.randint(22,50,n),
    'city': np.random.choice(['delhi','mumbai','pune','lucknoow','banglore'],n),
    'salary': np.random.randint(20000,55000,n),
    'performance': np.random.choice(['good','poor','very_good'],n)
})
```

```python
cust_data
```

```python
plot = sns.barplot(data=cust_data,x='department',y='salary')
```

```python
plt.figure(figsize=(11, 5))
sns.violinplot(
    data=cust_data,
    x='department',
    y='salary',
    palette='pastel',
    inner='box'
)
plt.title('Salary Distribution by Department (Violin Plot)', fontsize=16)
plt.xlabel('Department')
plt.ylabel('Salary (Rs)')
plt.tight_layout()
plt.show()
     
```

```python
plt.figure(figsize=(11, 5))
sns.boxplot(
    data=cust_data,
    x='department',
    y='salary',
    palette='Set2',
    width=0.5
)
plt.title('Salary Distribution by Department (Box Plot)', fontsize=16)
plt.xlabel('Department')
plt.ylabel('Salary (Rs)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
print('Box Plot shows: Median (line), Q1-Q3 (box), Whiskers, Outliers (dots)')
     
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(
    data=cust_data,
    x='salary',
    bins=30,
    kde=True,
    color='steelblue',
    edgecolor='white'
)
plt.axvline(cust_data['salary'].mean(), color='red', linestyle='--', label=f'Mean: Rs{cust_data["salary"].mean():,.0f}')
plt.title('salary Distribution', fontsize=16)
plt.xlabel('salary (Rs)')
plt.ylabel('Count')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(11, 5))
sns.stripplot(
    data=cust_data,
    x='department',
    y='salary',
    hue='gender',
    palette=['steelblue', 'tomato'],
    jitter=True,
    alpha=0.6,
    size=4
)
plt.title('Individual Salary Points by Department', fontsize=16)
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
     
```
